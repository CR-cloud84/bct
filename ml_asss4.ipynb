```python
# Implement Gradient Descent Algorithm to find the local minima of a function.
# For example, find the local minima of the function y=(x+3)2 starting from the point x=2.
```

```python
def gradient_descent(starting_x, learning_rate, iterations):
    x = starting_x
    
    for i in range(iterations):
        # Calculate the gradient: dy/dx = 2 * (x + 3)
        gradient = 2 * (x + 3)
        
        # Update x by moving against the gradient
        x = x - learning_rate * gradient
        
        # Optional: Print the progress every 5 iterations
        if i % 5 == 0:
            print(f"Iteration {i:2}: x = {x:.4f}, y = {(x+3)**2:.4f}")
            
    return x

# Set parameters
start_point = 2.0
alpha = 0.1
epochs = 50

final_x = gradient_descent(start_point, alpha, epochs)
print(f"\nLocal minima found at x = {final_x:.4f}")
```

```python
def gradient_descent(f, start_x, learning_rate=0.1, epochs=50, h=1e-5):
    x = start_x
    
    for i in range(epochs):
        # 1. Automatically approximate the derivative at the current x
        gradient = (f(x + h) - f(x - h)) / (2 * h)
        
        # 2. Step in the opposite direction of the gradient
        x = x - learning_rate * gradient
        
    return x

# --- Test Cases ---

# Test 1: Your original function y = (x + 3)^2
func_1 = lambda x: (x + 3)**2
min_1 = gradient_descent(func_1, start_x=2.0)
print(f"Minima 1: x = {min_1:.4f}")  # Expected: -3.0000

# Test 2: A different function y = x^2 - 4x + 4
func_2 = lambda x: x**2 - 4*x + 4
min_2 = gradient_descent(func_2, start_x=0.0)
print(f"Minima 2: x = {min_2:.4f}")  # Expected: 2.0000
```

```python

import numpy as np
import matplotlib.pyplot as plt

def gradient_descent(f, start_x, learning_rate=0.1,
                     epochs=50, h=1e-5):

    x = start_x
    history = [x]

    for i in range(epochs):
        gradient = (f(x+h) - f(x-h)) / (2*h)
        x = x - learning_rate * gradient
        history.append(x)

    return x, history


# Function
f = lambda x: (x + 3)**2

# Run gradient descent
minimum, history = gradient_descent(f, start_x=2)

# Plot function
x_values = np.linspace(-6, 5, 200)
y_values = f(x_values)

plt.plot(x_values, y_values, label="f(x) = (x+3)^2")

# Plot descent steps
for i, x in enumerate(history):
    plt.scatter(x, f(x), color="red")
    if i > 0:
        plt.plot(
            [history[i-1], x],
            [f(history[i-1]), f(x)],
            "r--"
        )

plt.scatter(minimum, f(minimum), color="green",
            s=100, label="Minimum")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent Visualization")
plt.legend()
plt.grid()
plt.show()

print("Minimum x:", minimum)
```
