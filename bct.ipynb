```python

// SPDX-License-Identifier: MIT
pragma solidity ^0.8.0;

contract StudentData {

    // Structure to store student information
    struct Student {
        uint id;
        string name;
        uint age;
        string course;
    }

    // Array to store multiple students
    Student[] public students;

    // Add student data
    function addStudent(
        uint _id,
        string memory _name,
        uint _age,
        string memory _course
    ) public {
        students.push(Student(_id, _name, _age, _course));
    }

    // Get student data
    function getStudent(uint index)
        public
        view
        returns (uint, string memory, uint, string memory)
    {
        Student memory s = students[index];

        return (s.id, s.name, s.age, s.course);
    }

    // Fallback function
    fallback() external {
    }
}
```

```python

// SPDX-License-Identifier: MIT
pragma solidity ^0.8.0;

contract MyBank
{
    // Stores balance of each customer's wallet address
    mapping(address => uint) private _balances;

    // Deposit money
    function deposit() public payable returns (uint)
    {
        require(msg.value > 0, "Deposit amount must be greater than zero");

        _balances[msg.sender] += msg.value;

        return _balances[msg.sender];
    }

    // Withdraw money
    function withdraw(uint withdrawAmount) public returns (uint)
    {
        require(withdrawAmount > 0, "Amount must be greater than zero");
        require(_balances[msg.sender] >= withdrawAmount, "Insufficient balance");

        _balances[msg.sender] -= withdrawAmount;

        payable(msg.sender).transfer(withdrawAmount);

        return _balances[msg.sender];
    }

    // Show balance
    function viewBalance() public view returns (uint)
    {
        return _balances[msg.sender];
    }
}
```
