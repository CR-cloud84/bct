```python
import time
import tracemalloc
import matplotlib.pyplot as plt


# ---------------------------------
# Recursive Fibonacci
# ---------------------------------
        
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n - 1) + fib_recursive(n - 2)


# ---------------------------------
# Iterative Fibonacci
# ---------------------------------

def fib_iterative(n):
    a = 0
    b = 1

    for i in range(n):
        a, b = b, a + b

    return a


# ---------------------------------
# Get maximum n
# ---------------------------------

max_n = int(input("Enter maximum n: "))


# Lists to store results
n_values = []
recursive_times = []
iterative_times = []

recursive_memory = []
iterative_memory = []


# ---------------------------------
# Test different values of n
# ---------------------------------

for n in range(1, max_n + 1):

    n_values.append(n)

    # -----------------------------
    # Recursive Fibonacci
    # -----------------------------

    tracemalloc.start()

    start = time.perf_counter()

    recursive_result = fib_recursive(n)

    recursive_time = time.perf_counter() - start

    current, recursive_mem = tracemalloc.get_traced_memory()

    tracemalloc.stop()


    # -----------------------------
    # Iterative Fibonacci
    # -----------------------------

    tracemalloc.start()

    start = time.perf_counter()

    iterative_result = fib_iterative(n)

    iterative_time = time.perf_counter() - start

    current, iterative_mem = tracemalloc.get_traced_memory()

    tracemalloc.stop()


    # Store results
    recursive_times.append(recursive_time)
    iterative_times.append(iterative_time)

    recursive_memory.append(recursive_mem)
    iterative_memory.append(iterative_mem)


# ---------------------------------
# Display Results
# ---------------------------------

print("\nResults")
print("--------------------------------")

print("n values:", n_values)

print("\nRecursive Execution Times:")
print(recursive_times)

print("\nIterative Execution Times:")
print(iterative_times)

print("\nRecursive Memory Usage:")
print(recursive_memory)

print("\nIterative Memory Usage:")
print(iterative_memory)


# ---------------------------------
# Time Complexity Information
# ---------------------------------

print("\nTime Complexity")
print("--------------------------------")
print("Recursive: O(2^n)")
print("Iterative: O(n)")


# ---------------------------------
# Space Complexity Information
# ---------------------------------

print("\nSpace Complexity")
print("--------------------------------")
print("Recursive: O(n)")
print("Iterative: O(1)")


# ---------------------------------
# Time Complexity Graph
# ---------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    n_values,
    recursive_times,
    marker="o",
    label="Recursive"
)

plt.plot(
    n_values,
    iterative_times,
    marker="o",
    label="Iterative"
)

plt.xlabel("Value of n")
plt.ylabel("Execution Time (seconds)")
plt.title("Recursive vs Iterative Fibonacci - Time Complexity")

plt.legend()
plt.grid(True)

plt.show()


# ---------------------------------
# Space Complexity Graph
# ---------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    n_values,
    recursive_memory,
    marker="o",
    label="Recursive"
)

plt.plot(
    n_values,
    iterative_memory,
    marker="o",
    label="Iterative"
)

plt.xlabel("Value of n")
plt.ylabel("Memory Usage (bytes)")
plt.title("Recursive vs Iterative Fibonacci - Space Complexity")

plt.legend()
plt.grid(True)

plt.show()
```
