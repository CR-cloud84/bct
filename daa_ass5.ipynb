```python

import time
import sys


def solve(n, first_row, first_col):
    board = [-1] * n

    board[first_row] = first_col

    def is_safe(row, col):
        for r in range(n):
            if board[r] == -1:
                continue

            c = board[r]

            if c == col or abs(c - col) == abs(r - row):
                return False

        return True

    def backtrack(row):

        if row == n:
            return True

        # Skip the row containing the already-placed queen
        if row == first_row:
            return backtrack(row + 1)

        for col in range(n):
            if is_safe(row, col):
                board[row] = col

                if backtrack(row + 1):
                    return True

                # Backtrack
                board[row] = -1

        return False

    # Start measuring time
    start_time = time.perf_counter()

    found = backtrack(0)

    # Stop measuring time
    end_time = time.perf_counter()

    execution_time = end_time - start_time

    # Space used by the board
    board_space = sys.getsizeof(board) + sum(
        sys.getsizeof(x) for x in board
    )

    print("\n-----------------------------")
    print("N-Queens Result")
    print("-----------------------------")

    print("N =", n)
    print("First Queen =", (first_row, first_col))

    if found:
        print("Solution found:")
        for row in range(n):
            for col in range(n):
                if board[row] == col:
                    print("Q", end=" ")
                else:
                    print(".", end=" ")
            print()
    else:
        print("No solution exists.")

    print("\nPerformance:")
    print("Theoretical Time Complexity: O(n × n!)")
    print("Theoretical Space Complexity: O(n)")
    print("Execution Time:", execution_time, "seconds")
    print("Board Space:", board_space, "bytes")

    return found, execution_time, board_space


# --------------------------------------------------
# Store results for different values of n
# --------------------------------------------------

n_values = []
execution_times = []
space_values = []

# Test different values of n
for n in range(4, 11):

    first_row = 0
    first_col = 1

    found, execution_time, board_space = solve(
        n,
        first_row,
        first_col
    )

    n_values.append(n)
    execution_times.append(execution_time)
    space_values.append(board_space)


# --------------------------------------------------
# Display stored results
# --------------------------------------------------

print("\n=============================")
print("Stored Results")
print("=============================")

print("N values:", n_values)
print("Execution times:", execution_times)
print("Space values:", space_values)
```

```python

import matplotlib.pyplot as plt


# --------------------------------------------------
# Plot Execution Time
# --------------------------------------------------

plt.figure()

plt.plot(n_values, execution_times, marker='o')

plt.xlabel("N")
plt.ylabel("Execution Time (seconds)")
plt.title("N-Queens: N vs Execution Time")

plt.grid(True)
plt.show()


# --------------------------------------------------
# Plot Space Usage
# --------------------------------------------------

plt.figure()

plt.plot(n_values, space_values, marker='o')

plt.xlabel("N")
plt.ylabel("Space Used (bytes)")
plt.title("N-Queens: N vs Space Usage")

plt.grid(True)
plt.show()
```
